## 3. Paper 2 backbone: DSConv

**What DSConv does.** It splits a standard 3×3 stride-2 convolution into two steps:
1. **Depthwise 3×3 conv, stride 2:** each channel is filtered separately and the resolution is halved.
2. **Pointwise 1×1 conv:** mixes the channels and sets the output channel count.

Both steps use BatchNorm + SiLU (ultralytics `Conv` blocks). This needs far fewer parameters and FLOPs than a full 3×3 conv.

**Where it sits in YOLO26n** (Stage 2, 3, 4 in Paper 2, Figure 6).

| Layer | Base YOLO26n | Paper 2 backbone | Output |
|---|---|---|---|
| 3 | Conv 3×3, stride 2 | **DSConv** | P3/8 (80 × 80) |
| 5 | Conv 3×3, stride 2 | **DSConv** | P4/16 (40 × 40) |
| 7 | Conv 3×3, stride 2 | **DSConv** | P5/32 (20 × 20) |

One Conv is replaced by one DSConv with the same input and output channels, so no layer index changes and the neck connections stay the same.
Only layers 3, 5 and 7 start from scratch; every other layer loads `yolo26n.pt`.

**Why it fits the task.** It makes the model lighter, which can raise the speed on a Raspberry Pi 5.

**Caution.** In Paper 2, DSConv alone lowered Precision from 92.2 % to 90.7 %. The key question of this experiment is how much
precision is lost for the speed gained.

```python
%%writefile dsconv_backbone.py
"""Paper 2 backbone for YOLO26n: stride-2 Conv at layers 3, 5, 7 -> DSConv (depthwise 3x3 s2 + pointwise 1x1)."""
import torch.nn as nn

from ultralytics.models.yolo.detect import DetectionTrainer
from ultralytics.nn.modules.conv import Conv
from ultralytics.nn.tasks import DetectionModel
from ultralytics.utils import RANK

DSCONV_LAYERS = (3, 5, 7)


class DSConv(nn.Module):
    """Depthwise separable convolution: depthwise kxk (stride s) + pointwise 1x1, each with BN + SiLU."""

    def __init__(self, c1, c2, k=3, s=2):
        super().__init__()
        self.dw = Conv(c1, c1, k, s, g=c1)
        self.pw = Conv(c1, c2, 1, 1)

    def forward(self, x):
        return self.pw(self.dw(x))


def add_dsconv(model, layers=DSCONV_LAYERS):
    """Replace the stride-2 Conv at the given backbone layers with DSConv (same in/out channels). Returns the model."""
    for i in layers:
        old = model.model[i]
        assert isinstance(old, Conv) and old.conv.kernel_size == (3, 3) and old.conv.stride == (2, 2), f"layer {i} is not a 3x3 s2 Conv"
        new = DSConv(old.conv.in_channels, old.conv.out_channels, 3, 2)
        new.i, new.f, new.type = old.i, old.f, "DSConv"  # keep the routing attributes used by the model's forward
        new.np = sum(p.numel() for p in new.parameters())
        model.model[i] = new
    return model


class DSConvTrainer(DetectionTrainer):
    """DetectionTrainer that inserts DSConv, then loads the pretrained weights (DSConv layers start from scratch)."""

    def get_model(self, cfg=None, weights=None, verbose=True):
        model = DetectionModel(cfg, nc=self.data["nc"], ch=self.data["channels"], verbose=False)
        add_dsconv(model)
        if weights:
            model.load(weights)
        if RANK in {-1, 0}:
            print(f"Paper 2 backbone: DSConv at layers {DSCONV_LAYERS}")
        if verbose and RANK in {-1, 0}:
            model.info()
        return model
```

```python
import importlib, dsconv_backbone
importlib.reload(dsconv_backbone)
from dsconv_backbone import add_dsconv, DSConvTrainer
from ultralytics.nn.tasks import DetectionModel, yaml_model_load
from ultralytics.utils.torch_utils import get_flops

def fused_params_gflops(model, imgsz=640):
    m = deepcopy(model).float().cpu().eval()
    m.fuse(verbose=False)
    return sum(p.numel() for p in m.parameters()) / 1e6, get_flops(m, imgsz)

# structure check (untrained): base vs Paper 2 backbone
base = DetectionModel(yaml_model_load("yolo26n.yaml"), nc=len(names), verbose=False)
dsconv = add_dsconv(DetectionModel(yaml_model_load("yolo26n.yaml"), nc=len(names), verbose=False))
dsconv.eval(); _ = dsconv(torch.zeros(1, 3, 640, 640))  # forward check
print("layers 3, 5, 7:", [type(dsconv.model[i]).__name__ for i in (3, 5, 7)])
for tag, m in (("base", base), ("Paper 2 backbone", dsconv)):
    pm, gf = fused_params_gflops(m)
    print(f"{tag:17s} Params {pm:.3f} M | GFLOPs {gf:.2f}")
```

## 4. Training

Only the Paper 2 backbone model is trained.

```python
from ultralytics import YOLO

EPOCHS  = 100
IMGSZ   = 640
BATCH   = 16
WORKERS = 2
SEED    = 0
PROJECT = str(WORK / "runs")

COMMON = dict(
    data=DATA_YAML, epochs=EPOCHS, imgsz=IMGSZ, batch=BATCH, seed=SEED, deterministic=True,
    pretrained="yolo26n.pt", device=0, workers=WORKERS, cache=False, patience=30, cos_lr=False,
    project=PROJECT, exist_ok=True, plots=True,
    # channels_last=False: ultralytics' MuSGD crashes on channels_last weights ("view size is not compatible").
    optimizer="MuSGD", lr0=0.01, momentum=0.9, channels_last=False,
)

# (row label in the table, run name)
VARIANTS = [
    ("DSConv Backbone (Paper 2)", "bb_dsconv"),
]

def best_path(run):
    return Path(PROJECT) / f"{run}_s{SEED}" / "weights" / "best.pt"

for label, run in VARIANTS:
    print(f"\n========== {label} ({run}) ==========")
    model = YOLO("yolo26n.yaml")
    model.train(trainer=DSConvTrainer, **dict(COMMON, name=f"{run}_s{SEED}"))
    del model; gc.collect(); torch.cuda.empty_cache()
```

## 5. Test-set evaluation: Precision, Recall, F1-score, mAP@50, Params, GFLOPs

Params and GFLOPs are counted on the fused inference model, the same way as the baseline row.

```python
import dsconv_backbone  # must be imported before loading a DSConv checkpoint

rows = []
for label, run in VARIANTS:
    w = best_path(run)
    if not w.exists():
        print(f"{run}: best.pt not found, train it first"); continue
    r = YOLO(str(w)).val(data=DATA_YAML, split="test", imgsz=IMGSZ, batch=16, device=0, plots=False, verbose=False)
    P, R = float(r.box.mp), float(r.box.mr)
    params, gflops = fused_params_gflops(YOLO(str(w)).model, IMGSZ)
    rows.append({"run": run, "Backbone Variant": label,
                 "Params (M)": round(params, 2), "GFLOPs": round(gflops, 2),
                 "Precision": round(P, 3), "Recall": round(R, 3),
                 "F1-score": round(2 * P * R / (P + R + 1e-9), 3),
                 "mAP@50": round(float(r.box.map50), 3)})
METRICS = pd.DataFrame(rows)
METRICS.drop(columns="run")
```

## 6. Package for Raspberry Pi 5 and Jetson

Creates `paper2_backbone_edge_bench.zip` in the **Output** tab with:

| Item | Used on |
|---|---|
| `bb_dsconv_ncnn_model/` (NCNN model, imgsz 640) | Raspberry Pi 5 |
| `bb_dsconv.pt` | Jetson |
| `bench_images/` (50 real test images, seed 0) | both |
| `dsconv_backbone.py` | Jetson (needed to load the DSConv `.pt`) |
| `pi5_bench.py`, `jetson_bench.py` | benchmark scripts |

Both scripts: batch 1, imgsz 640, cycle through the 50 real test images (loaded into memory first),
100 warm-up runs, then 1000 timed runs end-to-end (preprocessing + inference + postprocessing), and print the **median** ms and FPS.
Unzip into its own folder, e.g. `python -m zipfile -e paper2_backbone_edge_bench.zip paper2_backbone_edge_bench`.

```python
EDGE_IMGSZ = 640
N_BENCH_IMAGES = 50
EDGE = WORK / "paper2_backbone_edge_bench"
shutil.rmtree(EDGE, ignore_errors=True)
EDGE.mkdir()
shutil.copy("dsconv_backbone.py", EDGE / "dsconv_backbone.py")

# 50 real test images (fixed seed, so every model and device uses the same images)
test_imgs = sorted(p for p in (DATA_ROOT / "test" / "images").glob("*") if p.suffix.lower() in {".jpg", ".jpeg", ".png", ".bmp"})
random.Random(SEED).shuffle(test_imgs)
(EDGE / "bench_images").mkdir()
for p in test_imgs[:N_BENCH_IMAGES]:
    shutil.copy(p, EDGE / "bench_images" / p.name)
print(f"bench_images: {len(list((EDGE / 'bench_images').glob('*')))} real test images")

for _, run in VARIANTS:
    w = best_path(run)
    if not w.exists():
        print(f"{run}: best.pt not found, skipped"); continue
    shutil.copy(w, EDGE / f"{run}.pt")
    p = YOLO(str(w)).export(format="ncnn", imgsz=EDGE_IMGSZ, half=False)
    shutil.copytree(p, EDGE / f"{run}_ncnn_model")  # ultralytics needs the "_ncnn_model" suffix
    print(f"{run:12s} -> {run}.pt, {run}_ncnn_model/")

COMMON_BENCH = """
import glob, statistics, sys, time
import cv2

IMGSZ, WARMUP, RUNS = {imgsz}, 100, 1000
imgs = [cv2.imread(p) for p in sorted(glob.glob("bench_images/*"))]
imgs = [im for im in imgs if im is not None]
if not imgs:
    sys.exit("No images found in bench_images/. Run this script inside the unzipped folder.")
print(f"{{len(imgs)}} real test images, imgsz {{IMGSZ}}, batch 1, {{WARMUP}} warm-up + {{RUNS}} timed runs (median)")


def bench(model, **kw):
    for i in range(WARMUP):
        model.predict(imgs[i % len(imgs)], imgsz=IMGSZ, verbose=False, **kw)
    times = []
    for i in range(RUNS):
        t = time.perf_counter()
        model.predict(imgs[i % len(imgs)], imgsz=IMGSZ, verbose=False, **kw)
        times.append((time.perf_counter() - t) * 1000)
    return statistics.median(times)
""".format(imgsz=EDGE_IMGSZ)

PI5 = '''# Raspberry Pi 5 benchmark (NCNN). Run inside the unzipped folder:  python pi5_bench.py
from ultralytics import YOLO
''' + COMMON_BENCH + '''
models = sorted(glob.glob("*_ncnn_model"))
if not models:
    sys.exit("No *_ncnn_model folders found.")
for path in models:
    ms = bench(YOLO(path, task="detect"))
    print(f"{path.replace('_ncnn_model', ''):12s}  median {ms:8.2f} ms   {1000 / ms:7.2f} FPS")
'''

JETSON = '''# Jetson benchmark (PyTorch, CUDA FP16). Run inside the unzipped folder:  python jetson_bench.py
import sys
import torch

if not torch.cuda.is_available():
    sys.exit("WARNING: CUDA GPU not available. Install the NVIDIA Jetson PyTorch wheel for your JetPack, then run again.")
print("GPU:", torch.cuda.get_device_name(0))

sys.path.insert(0, ".")
import dsconv_backbone  # needed to load the DSConv checkpoint
from ultralytics import YOLO
''' + COMMON_BENCH + '''
models = sorted(glob.glob("*.pt"))
if not models:
    sys.exit("No .pt files found.")
for path in models:
    ms = bench(YOLO(path, task="detect"), device=0, half=True)
    print(f"{path.replace('.pt', ''):12s}  median {ms:8.2f} ms   {1000 / ms:7.2f} FPS")
'''

(EDGE / "pi5_bench.py").write_text(PI5)
(EDGE / "jetson_bench.py").write_text(JETSON)
shutil.make_archive(str(WORK / "paper2_backbone_edge_bench"), "zip", EDGE)
print("\nDownload paper2_backbone_edge_bench.zip from the Output tab.")
```
